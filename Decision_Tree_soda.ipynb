```python
import pandas as pd

data = {"Popcorn":[1,1,0,0,1,1,0],
       "Soda":[1,0,1,1,1,0,0],
       "Age":[7,12,18,35,38,50,83],
        "Movie" : [0,0,1,1,1,0,0]}
data

d=pd.DataFrame(data)
d
```

```python
x = d[["Popcorn","Soda","Age"]]   #feature
y = d ["Movie"]    #Prediction Target
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
model = DecisionTreeClassifier(criterion="entropy")
model.fit(x,y)
```

```python
prediction = [[1,1,38]]
result = model.predict(prediction)
result
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,3))
plot_tree(model,feature_names=["Popcorn","Soda","Age"],class_names=["yes","no"],filled=True)
plt.show()
```
