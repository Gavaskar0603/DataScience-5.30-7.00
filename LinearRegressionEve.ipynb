```python
e=pd.get_dummies(d["Gender"])
e
```

```python
d=pd.concat([d,e],axis=1)
d
```

```python
d.info()
```

```python
d=d.drop(["Gender"],axis=1)
d
```

```python
d["Years of Experience"]=d["Years of Experience"].fillna(d["Years of Experience"].mean())
d
```

```python
d.columns[d.isna().any()]
```

```python
d.info()
```

```python
d["Salary"]=d["Salary"].fillna(d["Salary"].mean())
d
```

```python
d.columns[d.isna().any()]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=d[["Years of Experience"]]
y=d["Salary"]

x_train,x_test,y_tarin,y_test = train_test_split(x,y,train_size=0.9,random_state=42)
x_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train,y_tarin)
```

```python
y_pred = lr.predict(x_test)
y_pred
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```
