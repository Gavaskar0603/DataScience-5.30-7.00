```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom as b
```

```python
n=10
p=0.5
x=np.arange(0,n+1)
pmf=b.pmf(x,n,p)
pmf
```

```python
plt.figure(figsize=(8,5))
plt.bar(x,pmf)
plt.title("Binomial Distribution")
plt.xlabel("Number of Success")
plt.ylabel("Probability")
plt.show()
```

```python
cdf = b.cdf(x,n,p)
cdf
```

```python
plt.figure(figsize=(8,5))
plt.plot(x,cdf,marker="o")
plt.title("Binomial Distribution")
plt.xlabel("Number of Success")
plt.ylabel("Probability")
plt.show()
```
