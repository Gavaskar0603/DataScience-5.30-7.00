```python
import numpy as np
```

```python
normal_dist = np.random.normal(loc=50,scale=3,size=100)
normal_dist
```

```python
nm=np.median(normal_dist)
nm
```

```python
n = np.array([1000,1500,2000])
na=np.concatenate([normal_dist,n])
na
```

```python
np.mean(na)
```

```python
np.median(na)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Normal_Dist":normal_dist})
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,7))
plt.subplot(1,2,1)
sns.histplot(normal_dist,kde=True)
plt.title("Normal dist")
plt.subplot(1,2,2)
sns.histplot(na,kde=True)
plt.title("Outliers")
plt.show()
```
